```python
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("twitter_dataset.csv")
print(df.shape)
print(df.head())
```

```python
x = df["Text"]
tweets = [r.lower().split() for r in x]
print(tweets[0])
```

```python
cbow = Word2Vec(
    sentences = tweets,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=10
)
print("Vocabulary size:",len(cbow.wv))
```

```python
tweet_embed = np.array([cbow.wv.get_mean_vector(r) for r in tweets])
print("Embedding shape:",tweet_embed.shape)
```

```python
user_input = input("Enter a topic or keywords:")
query_words = user_input.lower().split()
query_words = [w for w in query_words if w in cbow.wv]

if(len(query_words)==0):
    print("No matching words found.Try another keyword")
else:
    query_embeddings = cbow.wv.get_mean_vector(query_words)
    similarity = cosine_similarity([query_embeddings],tweet_embed)[0]

    top_indices = similarity.argsort()[-5:][::-1]

    for index in top_indices:
        print(df["Text"].iloc[index])
        print("Similarity score:",round(similarity[index],3))
        print("-" * 60)
```
