```python
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec
```

```python
df = pd.read_csv("twitter_dataset.csv")
print(df.head())
print(df.shape)
```

```python
x = df["Text"]
tweets = [r.lower().split() for r in x]
print(tweets[0])
```

```python
cbow = Word2Vec(
    sentences = tweets,
    min_count =2 ,
    window = 5,
    vector_size=100,
    sg=0,
    epochs=10
)
print("Vocabulary length:",len(cbow.wv))
```

```python
word = cbow.wv.index_to_key[0]
print("Word:",word)
print("Embeddings:",cbow.wv[word])
```

```python
print(cbow.wv.most_similar(word,topn=5))
```

```python
word1 = cbow.wv.index_to_key[0]
word2 = cbow.wv.index_to_key[1]
print("Similarity:",cbow.wv.similarity(word1,word2))
```

```python
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec

df = pd.read_csv("twitter_dataset.csv")
print(df.head())
print(df.shape)

x = df["Text"]
tweets = [r.lower().split() for r in x]
print(tweets[0])

cbow = Word2Vec(
    sentences = tweets,
    min_count =2 ,
    window = 5,
    vector_size=100,
    sg=0,
    epochs=10
)
print("Vocabulary length:",len(cbow.wv))

word = cbow.wv.index_to_key[0]
print("Word:",word)
print("Embeddings:",cbow.wv[word])

print(cbow.wv.most_similar(word,topn=5))


word1 = cbow.wv.index_to_key[0]
word2 = cbow.wv.index_to_key[1]
print("Similarity:",cbow.wv.similarity(word1,word2))
```
